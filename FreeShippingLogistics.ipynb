```python
import pandas as pd
```

```python
file_path = r'C:\Users\HP\Documents\Ecom Sales Report\Naxa Combo\Q4_24_SD_Naxa_Combo.xlsx'
```

```python
df = pd.read_excel(file_path)
```

# Data Cleaning

```python
df = df.drop(columns=['Delivery Date(DD/MM/YY)', 'Funds Remitted?', 'Source', 'Number of Naxa Bottles Sold'])
```

```python
df['Cost of Product'] = (df['Cost of Clovon Tea Per Sale(NGN)'] + df['Cost of Naxa Capsule Per Sale(NGN)'] + df['Cost of Teeth Whitening Per Sale(NGN)'])
```

```python
df = df.drop(columns=['Cost of Clovon Tea Per Sale(NGN)', 'Cost of Naxa Capsule Per Sale(NGN)', 'Cost of Teeth Whitening Per Sale(NGN)'])
```

# Feature Engineering

```python
df['Gross Profit'] = (df['Sale Value (NGN)'] - df['Cost of Product'])
```

```python
df['Net Profit'] = (df['Gross Profit'] - df['Cost of Delivery(NGN)'])
```

```python
#This measures the percentage of gross sales consumed by delivery expenses
df['% Cost of Delivery'] = ((df['Cost of Delivery(NGN)'] / df['Sale Value (NGN)']) * 100).round(2)
```

# Regional Margin Analysis

```python
 state_analysis = df.groupby('State').agg(
    total_revenue=('Sale Value (NGN)', 'sum'),
    total_gross_profit=('Net Profit', 'sum'),
    CoD_pct=('% Cost of Delivery', 'mean'),
    order_count=('Sale Value (NGN)', 'count')
).reset_index()
```

```python
# Calculate Regional Net Margin %
state_analysis['gross_margin_pct'] = ((state_analysis['total_gross_profit'] / state_analysis['total_revenue']) * 100).round(2)
```

```python
# Sort the data to highlight "Leakage" or "Profitability"
state_analysis = state_analysis.sort_values(by='CoD_pct', ascending=False)
```
